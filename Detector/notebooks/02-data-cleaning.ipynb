## Data Preprocessing

### 1. Load the libraries

```python
import pandas as pd
```

### 2. Load the dataset

```python
df = pd.read_csv(r'C:\Users\HP\Documents\PycharmProjects\network-anomaly-detector\Detector\data\UNSW-NB15.csv', low_memory=False)
pd.set_option('display.max_columns', None)
df.sample(15)
```

### 3. Datatypes of the features

```python
df[['sport', 'dsport']] = df[['sport', 'dsport']].apply(
    pd.to_numeric, errors='coerce'
)
df.info()
```

### 2. Finding abnormal values

```python
df['ct_ftp_cmd'].value_counts()
```

```python
df['service'].value_counts()
```

```python
df['attack_cat'].value_counts()
```

### 3. Handling abnormal values

```python
df['service']=df['service'].replace(['-'], pd.NA)
df['service']=df['service'].fillna('unknown')

df['ct_ftp_cmd'] = df['ct_ftp_cmd'].replace([' '], pd.NA)
df['ct_ftp_cmd'] = df['ct_ftp_cmd'].fillna(0)

df['attack_cat'] = df['attack_cat'].fillna('Normal')
df['attack_cat'] = df['attack_cat'].str.strip()
df["attack_cat"] = df["attack_cat"].replace({
    "Backdoors": "Backdoor"
})

print(df[['service','ct_ftp_cmd','attack_cat']].sample(25),"\n")
df['attack_cat'].value_counts()
```

### 4. Dropping unnecessary features

```python
data = df.copy()

data = data.drop(['srcip','dstip','dloss','sloss'],axis=1)
data.sample(10)
```

### 5. Drop Duplicates

```python
data = data.drop_duplicates()
print(data.duplicated().sum())
```

### 6. Sample of the cleaned data

```python
data.sample(15)
```

### 7. Save the dataset

```python
data.to_csv(r"C:\Users\HP\Documents\PycharmProjects\network-anomaly-detector\Detector\data\processed-data\clean-data.csv", index=False)
```
