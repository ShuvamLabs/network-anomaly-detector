### 9. Histograms

```python
columns = ["dur", "sbytes", "dbytes", "Sload", "Dload", "Spkts","Dpkts","res_bdy_len",'Sjit','Djit','Sintpkt','Dintpkt']

plt.figure(figsize=(15, 8))

for i, column in enumerate(columns):
    plt.subplot(4, 3, i + 1)
    plt.hist(np.log1p(df[column]), bins=50)
    plt.title(f"{column} - Log Transformed")

plt.tight_layout()
plt.show()
```

The features are is extremely right skewed, most of the values are near to 0 while some very large values extends the x axis suspected as outliers, log transformation has made the underlying distributions visible

### 11. Heatmap of most correlated columns

```python
top_features = (
    df.corr(numeric_only=True)["Label"].sort_values(key = abs , ascending = False).head(15).index
)

corr = df[top_features].corr()
plt.figure(figsize=(15,8))
sns.heatmap(corr, cmap='coolwarm', center =0, annot=True)
plt.show
```

### 12. Heatmap with least correlated features

```python
top_features = (
    df.corr(numeric_only=True)["Label"].sort_values(key = abs , ascending = False).tail(15).index
)

corr = df[top_features].corr()
plt.figure(figsize=(15,8))
sns.heatmap(corr, cmap='coolwarm', center =0, annot=True)
plt.show
```

### 13.All features heatmap

```python
plt.figure(figsize=(60,25))
sns.heatmap(df.corr(numeric_only = True), cmap='coolwarm', center =0, annot = True)
plt.show()
```

### 14. Scatter-Plots

```python
pairs = [
    ('sttl','ct_state_ttl'),
    ('ct_srv_src','ct_dst_ltm'),
    ('ct_src_dport_ltm','ct_dst_ltm'),
    ('dloss','dbytes')
]

sample = df.sample(25000)
fig, axes = plt.subplots(2, 2, figsize = (15,10))

for ax,(x,y) in zip(axes.flat , pairs):
    sns.scatterplot(
        data=sample,
        x=x,
        y=y,
        hue='Label',
        ax=ax
    )
    ax.set_title(f"{x} - {y}")
plt.tight_layout()
plt.show()
```

### 15. Box - PLot

```python
cols = ['dur','sbytes','dbytes','sloss','dloss','Sload','Dload','Spkts','Dpkts','trans_depth','res_bdy_len','Sjit','Djit','Sintpkt','Dintpkt']

df_log = np.log1p(df[cols])
df_log = pd.DataFrame(df_log, columns = cols)
df_log.plot(kind='box',subplots=True,layout=(3,5),figsize=(15,10))
plt.tight_layout()
plt.show()
```
